This notebook teaches how to visualize this metric with kafkanator framework, using the ADULT dataset.

```python
from sklearn.preprocessing import LabelEncoder
from kafkanator.dashboard import individual_fairness_plot
from kafkanator.dashboard import individual_fairness_dashboard
from kafkanator.dataviz.simfair2dplot  import SimilarityFairness2DPlot
from kafkanator.fairness.individual.metrics import individual_consistence
from kafkanator.dataviz.consistencyplot import ConsistencyPlot
from kafkanator.dataviz.preprocessor import ConsistencyFairnessPreprocessor
```

<h3> Data download </h3>

```python
from ucimlrepo import fetch_ucirepo 
# fetch dataset 
adult = fetch_ucirepo(id=2) 
# data (as pandas dataframes) 
data = adult.data.features 
y = adult.data.targets 
# Data cleaning
y['income'] = y['income'].apply(lambda x: x.replace('.',''))
data['sex'] = data['sex'].str.strip()
data['race'] = data['race'].str.strip()
data['marital-status'] = data['marital-status'].str.strip()
data['native-country'] = data['native-country'].str.strip()
```

<h3> The individual fairness dataviz method is defined as follows :  </h3>
<i> individual_fairness_dataviz ( data , target, plot_description, preprocessor_description, if_metric, sensitive_columns )</i>
Parameters : 
<ul>
    <li>data : The dataset you want to visualize </li>
    <li>target : The target or classification column </li>
    <li>plot_description : subclass of kafkanator.dataviz.creator.FairplotCreator</li>
    <li>preprocessor_description : subclass of kafkanator.dataviz.creator.IndividualFairnessPreprocessorCreator</li>
    <li>sensitive_columns : a sensitive column list. A subset of the dataset columns </li>
</ul>

```python
appli = individual_fairness_dataviz(data.iloc[0:1000],  y["income" ][0:1000], ConsistencyPlot() ,ConsistencyFairnessPreprocessor(),if_metric=individual_consistence , sensitive_columns=['race_encoded','sex_encoded'])
```

```python
from IPython.display import display, HTML
display(HTML("""
<style>
    div.output_scroll {height: 150em; }
</style>
"""))
if __name__ == '__main__':
    appli.run(debug=True, port=8050)
```
