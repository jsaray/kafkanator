```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
from kafkanator.dashboard import individual_fairness_dataviz
from kafkanator.fairness.individual.metrics import counterfactual_token_fairness_gap
from counterfactualplot import CounterfactualGapPlot
from conterfactualfairnesspreprocessor import ConterfactualFairnessPreprocessor
```

```python
compas_initial = pd.read_csv ( './compas_without_duplicates.csv' )
```

```python
class BiasedPredictor():
    def predict(self,X):
        predictions = []
        for a in X:
            if a[5] == 0 :
                predictions.append(1)
            elif a[5] == 1 :
                predictions.append(3)
            elif a[6] == 1 :
                predictions.append(2)
            elif a[6] == 0:
                predictions.append(3)
        return predictions
```

```python
bp = BiasedPredictor()
```

```python
compas_initial
```

```python
cf_preprocessor = ConterfactualFairnessPreprocessor()
appli = individual_fairness_dataviz(compas_initial.iloc[0:500,:] , compas_initial["ScoreText"].iloc[0:500] , CounterfactualGapPlot() ,cf_preprocessor,if_metric=counterfactual_token_fairness_gap,predictor=bp,sensitive_columns=['C_Ethnic_Code_Text','C_Sex_Code_Text'] )
```

```python
from IPython.display import display, HTML
display(HTML("""
<style>
    div.output_scroll {height: 150em; }
</style>
"""))
if __name__ == '__main__':
    appli.run(debug=True, port=8050)
```
